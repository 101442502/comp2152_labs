```python
# String for the name of gym member
gym_member = "Alex Alliton" 
# Float for preffered weight in kg
preferred_weight_kg = 20.5
# Integer for highest number of reps performed
highest_reps = 25
# Boolean for showing if the membership is active or not
membership_active = True
```

```python
# Dictionary: Stores workout stats with keys as friends name(string) and values as tuples of integer for yoga, running and weightlifting
workout_stats = {
    "Alex" : (30,40,50),
    "Jamie" : (40,50,40),
    "Taylor" : (50,50,10)}
```

```python
totals = {}

for friend, minutes in workout_stats.items():
    if isinstance(minutes, tuple):  # Ensure we only process tuples
        total_minutes = sum(minutes)  # Calculate total minutes
        totals[f"{friend}_Total"] = total_minutes  # Store in the temporary dictionary

workout_stats.update(totals)

# 2D List contains worskout minutes for each friend
workout_list = [list(minutes) for friend, minutes in workout_stats.items() if isinstance(minutes, tuple)]
```

```python
yoga_running_minutes = [row[:2] for row in workout_list]
print("Yoga & Running Minutes for All Friends:", yoga_running_minutes)

weightlifting_minutes_lasttwo = [row[2] for row in workout_list[-2:]]
print("Weightlifting Minutes for Last Two Friends:", weightlifting_minutes_last_two)
```

```python
for friend, minutes in workout_stats.items():
    if "_Total" in friend and minutes >= 120:  
        friend_name = friend.replace("_Total", "")
        print(f"Great job staying active, {friend_name}!")
```

```python
friend_name = input("Enter a friend's name: ")

if friend_name in workout_stats: 
    yoga, running, weightlifting = workout_stats[friend_name]
    total_minutes = workout_stats[f"{friend_name}_Total"] 
    print(f"{friend_name}'s workout details:")
    print(f"Yoga: {yoga} mins")
    print(f"Running: {running} mins")
    print(f"Weightlifting: {weightlifting} mins")
    print(f"Total Workout Time: {total_minutes} mins")
else:
    print(f"Friend {friend_name} not found in the records.")
```

```python
total_minutes_list = [(friend.replace("_Total", ""), minutes) for friend, minutes in workout_stats.items() if "_Total" in friend]

highest_friend = max(total_minutes_list, key=lambda x: x[1])
lowest_friend = min(total_minutes_list, key=lambda x: x[1])

print(f"The friend with the highest workout minutes is {highest_friend[0]} with {highest_friend[1]} minutes.")
print(f"The friend with the lowest workout minutes is {lowest_friend[0]} with {lowest_friend[1]} minutes.")
```
